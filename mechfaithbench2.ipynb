# MechFaithBench: устойчивость ER к способу закрашивания и к области

Ноутбук для Kaggle, **GPU T4 x2, Internet On**. Датасет MechFaithBench должен быть подключён через Add Input.

Эталонный опыт уже сделан: область закрашивается чёрным, контроль — кусок той же площади в другом месте.
Здесь тот же опыт повторяется с одним изменением за раз:

| Что меняется | Варианты |
|---|---|
| чем закрашивать | `gray` — серый; `mean` — средний цвет картинки; `blur` — размытая копия той же области; `noise` — случайный шум |
| что закрашивать | `pair_envelope` — один общий прямоугольник пары вместо объединения отдельных прямоугольников |

Контрольные куски во всех вариантах закрашивания одни и те же, поэтому варианты сравниваются на одних и тех же парах.

## Чекпоинты и восстановление

Все результаты лежат в `outputs/` внутри копии репозитория. Архив обновляется раз в 15 минут и после каждого шага.
При старте ноутбук ищет прошлые результаты в `/kaggle/input`: либо архив, либо уже распакованную папку `logits`.

```python
_last_zip = time.time()

def make_zip(note=""):
    global _last_zip
    tmp, n = ZIP_PATH + ".tmp", 0
    with zipfile.ZipFile(tmp, "w", zipfile.ZIP_DEFLATED) as z:
        for root, _, names in os.walk("outputs"):
            for name in names:
                z.write(os.path.join(root, name)); n += 1
    os.replace(tmp, ZIP_PATH); _last_zip = time.time()
    print(f"[checkpoint {note}] {n} files -> {ZIP_PATH} ({os.path.getsize(ZIP_PATH) / 1e6:.1f} MB)")

def find_previous(top):
    # (number of result files, kind, path) for every archive and every unpacked logits folder of this model
    found = []
    def add_zip(path):
        with zipfile.ZipFile(path) as z:
            found.append((sum(n.endswith(".parquet") for n in z.namelist()), "zip", path))
    if os.path.isfile(top):
        add_zip(top); return found
    for root, dirs, names in os.walk(top):
        for name in names:
            if name == ZIP_NAME:
                add_zip(os.path.join(root, name))
        if os.path.basename(root) == "logits" and os.path.isdir(os.path.join(root, MODEL, "clean")):
            found.append((len(glob.glob(os.path.join(root, "**", "*.parquet"), recursive=True)), "dir", root))
            dirs[:] = []
    return sorted(found)

def restore():
    if glob.glob("outputs/logits/**/*.parquet", recursive=True):
        print("outputs/ already has results -- continuing from them"); return
    found = find_previous(RESTORE_FROM or "/kaggle/input")
    if not found:
        print("no previous results found -- the clean pass and the reference experiment will be computed here"); return
    n, kind, path = found[-1]
    if kind == "zip":
        with zipfile.ZipFile(path) as z:
            z.extractall(".")
    else:
        shutil.copytree(path, "outputs/logits", dirs_exist_ok=True)
    print(f"restored from {path} ({kind}):", len(glob.glob("outputs/logits/**/*.parquet", recursive=True)), "result files")
    for d in sorted(glob.glob(f"outputs/logits/{MODEL}/*")):
        print("  ", os.path.basename(d), "|", len(glob.glob(d + "/part-*.parquet")), "parts")

def run_workers(script_args, name):
    # one process per GPU, each on its own share of the pairs; prints progress and archives periodically
    os.makedirs("outputs/logs", exist_ok=True)
    procs = []
    for i in range(N_GPUS):
        env = dict(os.environ, CUDA_VISIBLE_DEVICES=str(i), PYTHONUNBUFFERED="1", TOKENIZERS_PARALLELISM="false")
        log = open(f"outputs/logs/{name}_gpu{i}.log", "w")
        procs.append((subprocess.Popen([sys.executable, "-u"] + script_args + ["--shard", f"{i}/{N_GPUS}"], stdout=log, stderr=subprocess.STDOUT, env=env), log))
    last = [""] * N_GPUS
    def progress():
        for i in range(N_GPUS):
            lines = [l for l in open(f"outputs/logs/{name}_gpu{i}.log", errors="ignore").read().splitlines()
                     if " pairs" in l or "nothing left" in l or "copied from" in l]
            if lines and lines[-1] != last[i]:
                last[i] = lines[-1]; print(f"gpu{i}: {lines[-1].strip()}")
    while any(p.poll() is None for p, _ in procs):
        time.sleep(2 if FAKE else 20)
        progress()
        if time.time() - _last_zip > CHECKPOINT_MINUTES * 60:
            make_zip("periodic")
    progress()
    for i, (p, log) in enumerate(procs):
        log.close()
        if p.returncode != 0:
            print(f"--- gpu{i} failed, end of its log:\n" + "".join(open(f"outputs/logs/{name}_gpu{i}.log", errors="ignore").readlines()[-25:]))
            raise RuntimeError(f"{name}: worker {i} exited with code {p.returncode}")
    make_zip(name)

def run(script_args):
    r = subprocess.run([sys.executable] + script_args, capture_output=True, text=True)
    print(r.stdout[-20000:])
    if r.returncode != 0:
        print(r.stderr[-3000:]); raise RuntimeError(" ".join(script_args))

restore()
```

## 2. Эталон и варианты

Для каждой верно решённой пары: закрашенная область и `K` контрольных кусков, на обеих картинках — 2 × (1 + K) проходов на пару.
Сначала эталон (если он восстановлен из архива, пропускается), затем варианты по списку `EXPERIMENTS`.

Для второй области (`pair_envelope`) пары, у которых общий прямоугольник совпадает с объединением попиксельно, не пересчитываются:
их строки копируются из эталона. Считаются только пары, где области действительно разные.

```python
def run_experiment(operator, support):
    name = f"{operator}__{support}"
    args = ["scripts/intervene.py", "--model", MODEL, "--operator", operator, "--support", support, "--k", str(K)] + LIMIT
    if support != REFERENCE[1]:
        args += ["--reuse-support", REFERENCE[1]]
    print(f"\n######## {name} ########")
    t0 = time.time()
    run_workers(args, name)
    run(["scripts/analyze_er.py", "--model", MODEL, "--experiment", name + SUFFIX])
    print(f"{name}: {(time.time() - t0) / 60:.0f} min")

for operator, support in [REFERENCE] + [e for e in EXPERIMENTS if tuple(e) != REFERENCE]:
    run_experiment(operator, support)
```

Как читать таблицы ниже.

**Таблица 1** — каждый вариант на своих парах.
`er` — основная метрика: на сколько закрашивание области снижает вероятность правильного ответа сильнее, чем закрашивание контроля.
`er_yes`, `er_no` — то же отдельно для картинки с ответом «да» и с ответом «нет».
`roa` — доля пар, где закрашивание области вреднее каждого из контролей.

**Таблица 2** — разница с эталоном на общих парах, с 95% интервалом. Если интервал не накрывает ноль, разница не случайна.
`spearman_er_yes` — насколько совпадает порядок пар по величине эффекта (1 — полностью).

**Таблицы 3** — по категориям на парах, общих для всех вариантов; в конце — насколько совпадает порядок категорий с эталоном.

**Таблица 4** — вопросы про серый, серебристый и чёрный цвет. Серая или чёрная заливка сама делает область такого цвета, и эти вопросы показаны отдельно.
